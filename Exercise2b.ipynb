## Building Different Regression Models

Find the best regression model

16. Set independent and dependent variables

```python
X = df[['Miles Traveled', 'Number of Deliveries', 'Gas Price']]
Y = df['Travel Time']
```

17. Build and train the model

```python
# Add a constant term to the independent variables (required for statsmodels OLS)
X_sm = sm.add_constant(X)

# Build and train the Ordinary Least Squares (OLS) regression model
model = sm.OLS(Y, X_sm).fit()
```

18. Display the statistical summary

```python
print(model.summary())
```

19. Display the $R^2$

```python
print(f"R-squared (R^2): {model.rsquared:.4f}")
```
